```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df = pd.read_csv("/content/student_exam_data_v1.csv")
df
```

```python
plt.scatter(df.study_hours, df.exam_score)
plt.xlabel('study_hours', color= 'red')
plt.ylabel('exam_score', color= 'red')
plt.title('study_hours & exam_score', color='red')
plt.show()
```

```python
lr = linear_model.LinearRegression()
lr.fit(df[['study_hours']], df.exam_score)
```

```python
lr.score(df[['study_hours']], df.exam_score)
```

```python
lr.predict([[9.0]])
```

```python
lr.score(df[['study_hours']], df.exam_score)
lr.coef_
```

```python
lr.intercept_
```

```python
#    [y=wx + b] formula
predicted = 6.57735876 * 9.0 + 29.081508691186237
print(predicted)
```

```python
plt.scatter(df.study_hours, df.exam_score)
plt.xlabel('study_hours', color= 'red')
plt.ylabel('exam_score', color= 'red')
plt.title('study_hours & exam_score', color='red')
plt.plot(df.study_hours, lr.predict(df[['study_hours']]),color='red')
plt.show()
```
